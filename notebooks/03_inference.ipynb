# 03 · Inference
Load the packaged model directly, then call the running API.

```python
import json
import urllib.request

from heart import config
from heart.predict import HeartModel

model = HeartModel.load()
print(model.metadata["model_name"], "threshold", model.threshold)
```

```python
patient = json.loads((config.PROJECT_ROOT / "sample_request.json").read_text())
model.predict([patient, {**patient, "ca": None, "thal": None}])
```

## Call the API
Start it first: `uvicorn api.main:app --port 8000` or `docker run -p 8000:8000 heart-risk-api`.

```python
API = "http://localhost:8000"
req = urllib.request.Request(f"{API}/predict", data=json.dumps(patient).encode(),
                             headers={"Content-Type": "application/json"})
with urllib.request.urlopen(req) as resp:
    print(json.dumps(json.load(resp), indent=2))
```
